```python
import gradio as gr
from sklearn.datasets import load_wine
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import base64
from io import BytesIO

# 데이터 로드 및 초기화 함수
def load_and_preprocess():
    wine = load_wine()
    df = pd.DataFrame(wine.data, columns=wine.feature_names)
    df['target'] = wine.target
    return df, wine.feature_names

# 상관계수 히트맵 시각화 함수
def create_heatmap(df, columns):
    plt.figure(figsize=(12, 10))
    corr = df[columns].corr()
    sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
    plt.title("화학 성분 간 상관관계 분석", fontsize=15)

    buffer = BytesIO()
    plt.savefig(buffer, format='png', bbox_inches='tight')
    buffer.seek(0)
    img_base64 = base64.b64encode(buffer.read()).decode('utf-8')
    plt.close()
    return f"<img src='data:image/png;base64,{img_base64}' style='width:100%;'>"

# 변수 간 분포 산점도 함수
def create_scatter(df, x_col, y_col):
    plt.figure(figsize=(10, 8))
    sns.scatterplot(data=df, x=x_col, y=y_col, hue='target', palette='viridis', s=70)
    plt.title(f"와인 클래스별 분포 확인: {x_col} vs {y_col}", fontsize=14)

    buffer = BytesIO()
    plt.savefig(buffer, format='png', bbox_inches='tight')
    buffer.seek(0)
    img_base64 = base64.b64encode(buffer.read()).decode('utf-8')
    plt.close()
    return f"<img src='data:image/png;base64,{img_base64}' style='width:100%;'>"

# 대시보드 통합 실행 함수
def wine_analysis_dashboard(rows, heatmap_cols, scatter_x, scatter_y):
    df, feature_names = load_and_preprocess()

    df_preview = df.head(int(rows))
    heatmap_html = create_heatmap(df, heatmap_cols)
    scatter_html = create_scatter(df, scatter_y, scatter_x) # scatter_x와 scatter_y의 위치를 변경했습니다.

    return df_preview, heatmap_html, scatter_html

# 입력 옵션 데이터 로드
df_init, feature_names = load_and_preprocess()

# Gradio 인터페이스 구성
with gr.Blocks(theme=gr.themes.Default()) as demo:
    gr.Markdown("# ✶와인 분석 인터랙티브 대시보드✶")
    gr.Markdown("이 대시보드는 Scikit-learn의 load_wine 데이터셋을 분석하기 위해 제작되었습니다.")

    with gr.Row():
        with gr.Column(scale=1):
            rows_input = gr.Slider(minimum=5, maximum=178, value=10, step=1, label="데이터 조회 행 수")
            heatmap_cols_input = gr.CheckboxGroup(choices=feature_names, value=feature_names[:6], label="상관관계 분석 변수 선택")
            scatter_x_input = gr.Dropdown(choices=feature_names, value='alcohol', label="산점도 X축 항목")
            scatter_y_input = gr.Dropdown(choices=feature_names, value='color_intensity', label="산점도 Y축 항목")
            submit_btn = gr.Button("데이터 분석 실행", variant="primary")

        with gr.Column(scale=3):
            with gr.Tab("데이터 목록"):
                data_output = gr.Dataframe()
            with gr.Tab("상관관계 히트맵"):
                heatmap_output = gr.HTML()
            with gr.Tab("클래스 분포 산점도"):
                scatter_output = gr.HTML()

    submit_btn.click(
        fn=wine_analysis_dashboard,
        inputs=[rows_input, heatmap_cols_input, scatter_x_input, scatter_y_input],
        outputs=[data_output, heatmap_output, scatter_output]
    )

demo.launch(share=True)
```
